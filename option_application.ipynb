```python
import numpy as np
from numpy.polynomial import laguerre
from quantlib_sacha.quantlib.instruments.option import Option

import seaborn as sns
```

```python
spot = 100  # spot price
strike =100 # strike price
rf = 0.03  # risk-free rate
div_rate = 0.04 # dividend rate
volatility = 0.20 # volatility
maturity = 1 # time to maturity in years
IsCall = True # True for call, False for put
IsEuropean = True  # True for European's option, False for American's option
option_type = "VANILLA" #["VANILLA", "AVERAGE_STRIKE", "AVERAGE_PRICE", "STRIKE_MIN", "STRIKE_MAX", "DOWN_IN", "DOWN_OUT", "UP_IN", "UP_OUT"]
barrier = 100
div = None
```

# 1. European Option

```python
help(Option)
```

```python
IsEuropean = True
option_vanilla = Option(spot, strike, rf, div_rate, volatility, maturity, IsCall, IsEuropean, option_type, barrier, div)
IsEuropean = False
option_american = Option(spot, strike, rf, div_rate, volatility, maturity, IsCall, IsEuropean, option_type, barrier, div)
```

```python
price_vanilla = option_vanilla.price()
price_american = option_american.price(5, 2)
```

```python
price_vanilla = option_vanilla.price()
print("Price : ", price_vanilla)
delta = option_vanilla.delta()
print("Delta : ", delta)
gamma = option_vanilla.gamma()
print("Gamma: ", gamma)
theta = option_vanilla.theta()
print("Theta: ", theta)
rho = option_vanilla.rho()
print("rho: ", rho)
vega = option_vanilla.vega()
print("vega: ", vega)
```

```python
price = option_american.price(1000,50)
print("Price : ", price)
delta = option_american.delta(1000,50)
print("Delta : ", delta)
gamma = option_american.gamma(1000,50)
print("Gamma: ", gamma)
theta = option_american.theta(1000,50)
print("Theta: ", theta)
rho = option_american.rho(1000,50)
print("rho: ", rho)
vega = option_american.vega(1000,50)
print("vega: ", vega)
```
